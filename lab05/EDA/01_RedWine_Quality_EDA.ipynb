# LAB 1: PHÂN TÍCH KHÁM PHÁ DỮ LIỆU RƯỢU ĐỎ (RED WINE QUALITY)


```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

##  Đọc dữ liệu từ thư mục data/

```python
wine_df = pd.read_csv('../data/winequality-red.csv')
```

## Thống kê mô tả

```python
print("=== THỐNG KÊ MÔ TẢ RED WINE QUALITY ===")
display(wine_df.describe().T[['mean', '50%', 'std', 'min', 'max']])
```

## Trực quan hóa phân bố Quality

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=wine_df, x='quality', palette='Set2')
plt.title("Phân bố Chất lượng Rượu Đỏ (Quality)", fontsize=14)
plt.xlabel("Chất lượng (Quality)")
plt.ylabel("Số lượng mẫu")
plt.tight_layout()
plt.savefig('../reports/figures/wine_quality_dist.png', dpi=300)
plt.show()
```

## Ma trận tương quan Pearson

```python
plt.figure(figsize=(10, 8))
sns.heatmap(wine_df.corr(), annot=True, fmt=".2f", cmap='coolwarm', vmin=-1, vmax=1)
plt.title("Ma trận Tương quan - Red Wine Quality", fontsize=14)
plt.tight_layout()
plt.savefig('../reports/figures/wine_corr_matrix.png', dpi=300)
plt.show()
```

### GIẢI THÍCH Ý NGHĨA KẾT QUẢ CHẠY:
1. **Phân bố Quality:** Chất lượng rượu tập trung chủ yếu ở mức 5 và 6. Mức chất lượng thấp (3, 4) và cao (7, 8) chiếm tỷ lệ rất nhỏ, cho thấy tập dữ liệu bị mất cân bằng lớp (class imbalance).
2. **Ma trận Tương quan:** 
   - `alcohol` (nồng độ cồn) có tương quan thuận mạnh nhất với chất lượng rượu (`quality`) với hệ số khoảng +0.48.
   - `volatile acidity` (axit bay hơi) có tương quan âm (-0.39) với chất lượng rượu; nghĩa là nồng độ axit bay hơi càng cao thì rượu càng bị chua gắt và giảm chất lượng.
