# LAB 1: PHÂN TÍCH KHÁM PHÁ DỮ LIỆU TIỂU ĐƯỜNG (PIMA INDIANS DIABETES)

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Đọc dữ liệu

```python
diabetes_df = pd.read_csv('../data/diabetes.csv')
```

## Thống kê mô tả

```python
print("=== THỐNG KÊ MÔ TẢ PIMA INDIANS DIABETES ===")
display(diabetes_df.describe().T[['mean', '50%', 'std', 'min', 'max']])
```

## Biểu đồ Boxplot phát hiện Outliers

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=diabetes_df[['Glucose', 'BloodPressure', 'BMI', 'Age']], palette='Set3')
plt.title("Kiểm tra Outliers các chỉ số sức khỏe chính", fontsize=14)
plt.tight_layout()
plt.savefig('../reports/figures/diabetes_boxplot.png', dpi=300)
plt.show()
```

## Biểu đồ Scatter Glucose vs BMI theo Outcome

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=diabetes_df, x='Glucose', y='BMI', hue='Outcome', alpha=0.8, palette='Set1')
plt.title("Mối quan hệ giữa Glucose và BMI phân loại theo Outcome", fontsize=14)
plt.tight_layout()
plt.savefig('../reports/figures/diabetes_scatter.png', dpi=300)
plt.show()
```

###  GIẢI THÍCH Ý NGHĨA KẾT QUẢ CHẠY:
1. **Giá trị bất thường (Outliers & Missing):** Trong thống kê, các giá trị `Glucose`, `BloodPressure`, `BMI` bằng 0 là phi lý trong thực tế y khoa. Đây chính là các giá trị thiếu (missing values) được mã hóa thành 0.
2. **Mối quan hệ Glucose & BMI:** Những người có nhãn `Outcome = 1` (mắc bệnh tiểu đường) thường phân bố ở vùng `Glucose > 120` và `BMI > 30`. Điều này khẳng định Glucose và BMI là 2 đặc trưng quan trọng nhất để phân loại nguy cơ tiểu đường.
